## Creating the database and the tables and insertion of the data

```python
import pandas as pd
import mysql.connector

# ✅ Load your datasets (you already have them loaded, but including for completeness)
products = pd.read_excel("../data/products.xlsx")
users = pd.read_excel("../data/users.xlsx")
reviews = pd.read_excel("../data/reviews.xlsx")

# ✅ Connect to MySQL (initial connection to create DB)
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root"
)
cursor = conn.cursor()

# ✅ Create database if not exists
cursor.execute("CREATE DATABASE IF NOT EXISTS ecofriendlypr")
print("✅ Database 'ecofriendlypr' created or already exists.")
cursor.close()
conn.close()

# ✅ Reconnect to the created database
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root",
    database="ecofriendlypr"
)
cursor = conn.cursor()

# ✅ Create products table
cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(255),
    brand VARCHAR(100),
    category VARCHAR(100),
    price FLOAT,
    availability VARCHAR(50)
)
""")

# ✅ Create users table
cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    user_id INT,
    product_id INT,
    FOREIGN KEY (product_id) REFERENCES products(product_id)
)
""")

# ✅ Create reviews table
cursor.execute("""
CREATE TABLE IF NOT EXISTS reviews (
    product_name VARCHAR(255),
    user_id INT,
    ratings FLOAT,
    unique_reviews TEXT
)
""")

conn.commit()
print("✅ Tables created successfully.")

# ✅ Insert into products table
for _, row in products.iterrows():
    cursor.execute("""
        INSERT INTO products (product_id, product_name, brand, category, price, availability)
        VALUES (%s, %s, %s, %s, %s, %s)
    """, (
        int(row["Product ID"]),
        str(row["Product Name"]),
        str(row["Brand"]),
        str(row["Category"]),
        float(row["Price"]),
        str(row["Availability"])
    ))

# ✅ Insert into users table (typecasting to native types)
for _, row in users.iterrows():
    cursor.execute("""
        INSERT INTO users (user_id, product_id)
        VALUES (%s, %s)
    """, (
        int(row["User-ID"]),
        int(row["product_id"])
    ))

# ✅ Insert into reviews table (typecasting to native types)
for _, row in reviews.iterrows():
    cursor.execute("""
        INSERT INTO reviews (product_name, user_id, ratings, unique_reviews)
        VALUES (%s, %s, %s, %s)
    """, (
        str(row["Product Name"]),
        int(row["User-ID"]),
        float(row["Ratings"]),
        str(row["Unique Reviews"])
    ))

# ✅ Commit changes and close connection
conn.commit()
print("✅ Data inserted successfully into all tables.")

cursor.close()
conn.close()
print("🔗 MySQL connection closed.")
```
