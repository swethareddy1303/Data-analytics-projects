```python
import mysql.connector
import pandas as pd

# Database connection details
DB_HOST = "localhost"
DB_USER = "root"
DB_PASSWORD = "root"
DB_NAME = "eco_friendly"
TABLE_NAME = "product"

# Connect to MySQL
connection = mysql.connector.connect(
    host=DB_HOST,
    user=DB_USER,
    password=DB_PASSWORD,
    database=DB_NAME
)
cursor = connection.cursor()

# Create table
create_table_query = f"""
CREATE TABLE IF NOT EXISTS {TABLE_NAME} (
    product_id INT AUTO_INCREMENT PRIMARY KEY,
    user_id INT,
    product_name VARCHAR(255),
    category VARCHAR(100),
    material VARCHAR(255),
    price VARCHAR(50),
    description TEXT,
    brand VARCHAR(100),
    availability VARCHAR(50),
    ratings FLOAT
);
"""
cursor.execute(create_table_query)
connection.commit()
print("✅ Table created successfully!")

# Load data from Excel
file_path = "EcoFriendly_Products.xlsx"  # Ensure correct path
df = pd.read_excel(file_path)

# 🔹 Print actual column names in the dataset
print("🛠 Checking column names:", df.columns)

# 🔹 Rename "User-ID" to "User_ID" for consistency
df.rename(columns={"User-ID": "User_ID"}, inplace=True)

# Expected columns list
expected_columns = ["User_ID", "Product Name", "Category", "Material", "Price", "Description", "Brand", "Availability", "Ratings"]

# 🔹 Check for missing columns
missing_columns = [col for col in expected_columns if col not in df.columns]
if missing_columns:
    print(f"⚠️ Missing columns in file: {missing_columns}")
    exit()

# Insert data into the database
insert_query = f"""
INSERT INTO {TABLE_NAME} (user_id, product_name, category, material, price, description, brand, availability, ratings)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s);
"""

# 🔹 Insert data row by row
for _, row in df.iterrows():
    cursor.execute(insert_query, (
        row["User_ID"], row["Product Name"], row["Category"], row["Material"], row["Price"],
        row["Description"], row["Brand"], row["Availability"], row["Ratings"]
    ))

# Commit changes
connection.commit()
print("✅ Data successfully inserted into the database!")

# Close the connection
cursor.close()
connection.close()
print("🔗 MySQL connection closed.")
```
