## using statistical methods

```python

import pandas as pd
import numpy as np


# --- IQR Outlier Detection for Entire Dataset with Count ---
def detect_outliers_iqr(df):
    outliers = pd.DataFrame()
    outlier_counts = {}
    
    for column in df.select_dtypes(include=[np.number]):  # Loop through numeric columns
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers_column = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
        
        # Count the number of outliers in each column
        outlier_counts[column] = len(outliers_column)
        
        if not outliers_column.empty:
            outliers = pd.concat([outliers, outliers_column])
    
 
    
    print("Outliers detected using IQR:\n", outliers)
    print("\nOutlier Counts by Column:\n", outlier_counts)

# Detect Outliers
detect_outliers_iqr(df)



```

```python

# --- Function to Detect and Replace Outliers with Count Tracking ---
def replace_outliers_iqr_with_count(df):
    outlier_count_before = {}
    outlier_count_after = {}
    
    for column in df.select_dtypes(include=[np.number]):  # Loop through numeric columns
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Count outliers before replacement
        outlier_count_before[column] = len(df[(df[column] < lower_bound) | (df[column] > upper_bound)])
        
        # Replace outliers below lower bound
        df[column] = np.where(df[column] < lower_bound, lower_bound, df[column])
        
        # Replace outliers above upper bound
        df[column] = np.where(df[column] > upper_bound, upper_bound, df[column])
        
        # Count outliers after replacement (should be 0 if all outliers are replaced)
        outlier_count_after[column] = len(df[(df[column] < lower_bound) | (df[column] > upper_bound)])
    
    return df, outlier_count_before, outlier_count_after

# Replace Outliers and get counts
df_cleaned, outlier_count_before, outlier_count_after = replace_outliers_iqr_with_count(df)

print("Outlier Count Before Replacement:\n", outlier_count_before)
print("\nOutlier Count After Replacement:\n", outlier_count_after)
print("\nCleaned Dataset (Outliers Replaced):\n", df_cleaned)
```

## label encoding

```python
#PERFROMING LABEL ENCODING ON STATUS,PRODUCTLINE AND DEALSIZE


# --- Label Encoding Function for Specific Columns ---
def label_encode_selected_columns(df, columns):
    label_encoder = LabelEncoder()
    
    for column in columns:  # Loop through the specified columns
        df[column] = label_encoder.fit_transform(df[column])
    
    return df

# Specify columns to encode
columns_to_encode = ['STATUS', 'PRODUCTLINE', 'DEALSIZE']

# Apply Label Encoding
df = label_encode_selected_columns(df, columns_to_encode)

print("Dataset After Label Encoding:\n", df)


```
